Seaborn 

```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# load a dataset from seaborn
df = sns.load_dataset("penguins").dropna()

# set up correlation matrix
corr_matrix = df.select_dtypes(include=[np.number]).corr()

# figure size - size of graphic in inches
plt.figure(figsize=(8,6))

sns.heatmap(
    corr_matrix,
    annot=True
)

#title
plt.title("How Correlated are Different Features of Penguins?")
plt.show()
```

Plotly

```python
import plotly.express as px


# load plotly dataset
df = px.data.tips()

fig = px.scatter(
    df, 
    x="total_bill",
    y="tip",
    size="size",
    facet_col="time",
    facet_row="sex",
    color="smoker",
    title="Is There a Relationship Between Tips and Total Bill?",
    hover_data=["day", "time", "smoker"]
)

# resize your figure
fig.update_layout(
    width=500,
    height=500
)

fig.show()
```
