Matplotlib

```python
import matplotlib.pyplot as plt
import numpy as np

sample_data = np.linspace(0, 10, 100)
y1 = np.sin(sample_data)
y2 = np.cos(sample_data)
```

```python
# build the visuals

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5)) # create a container
fig.set_facecolor("pink")


# plot 1: sin
ax1.plot(sample_data, y1, color="green", linewidth=5, label="Sin")
ax1.set_title("Sin Wave", fontsize=12)
ax1.set_xlabel("Time")
ax1.set_ylabel("Amplitude")
ax1.grid(True, linestyle="-.")
ax1.legend(frameon=True) #adds a legend

# plot 2: cos
ax2.plot(sample_data, y2, color="orange", linewidth=5, linestyle="-.", label="Cos")
ax2.set_title("Cos Wave", fontsize=12)
ax2.set_xlabel("Time")
ax2.set_ylabel("Amplitude")
ax2.grid(True, linestyle="-.")
ax2.legend(frameon=True) #adds a legend

# supertitle
fig.suptitle("Sin and Cos Waves of Set Interval Sample Data", fontweight="bold")

plt.tight_layout()
plt.show()
```

Seaborn

```python
import seaborn as sns

# load in built in seaborn dataset
df = sns.load_dataset("tips")
# set the style
sns.set_theme(style="darkgrid")

fig, axes = plt.subplots(1, 2, figsize=(12, 5)) #set up container
# plot 1: Histogram
sns.histplot(
    data=df,
    x="total_bill",
    hue="time",
    kde=True,
    multiple="stack",
    palette="pastel",
    ax=axes[0] #places it on the left
)

# plot 2: box plot
sns.boxplot(
    data=df,
    x="day",
    y="total_bill",
    hue="sex",
    palette="colorblind",
    ax=axes[1]
)

plt.show()
```
